## Exercise 1 — Defining the Problem and Data Collection for Loan Default Prediction

**Problem Statement:**
Build a machine learning model that predicts whether a loan applicant is likely to default (fail to repay) on a loan, based on their financial and personal profile. The goal is to help a lender decide whether to approve a loan application and at what risk level, before money is lent out.

**Data types needed:**
- **Applicant personal details** — age, marital status, number of dependents, education level, employment status
- **Financial information** — income, existing debts, credit score, bank account history
- **Loan details** — requested loan amount, loan term, interest rate, purpose of the loan
- **Repayment history** — past loans taken and whether they were repaid on time, late, or defaulted

**Data sources:**
- **The financial institution's own internal records** — past loan applications, approvals, and repayment outcomes (this is the most valuable source, since it directly captures real default behavior)
- **Credit bureaus** — third-party credit scores and credit history reports
- **Applicant-submitted forms** — income statements, employment verification, ID documents collected at the time of application

## Exercise 2 — Feature Selection and Model Choice for Loan Default Prediction

**Dataset:** Loan Prediction Problem Dataset (Kaggle) — columns include Gender, Married, Dependents, Education, Self_Employed, ApplicantIncome, CoapplicantIncome, LoanAmount, Loan_Amount_Term, Credit_History, Property_Area, Loan_Status.

**Most relevant features and justification:**

- **Credit_History** — likely the single strongest predictor. A person's past track record of repaying credit is usually the best available signal for whether they'll repay in the future — this mirrors exactly what we found with RAM being the dominant predictor in the Mobile Price dataset two weeks ago; one feature often dominates.
- **ApplicantIncome + CoapplicantIncome** — combined household income directly determines repayment capacity relative to the loan size.
- **LoanAmount** relative to income — a large loan against a small income is inherently riskier; this suggests a derived feature (loan-to-income ratio) could be even more useful than either raw column alone, similar to how we engineered `FamilySize` from raw columns on the Titanic dataset.
- **Loan_Amount_Term** — the repayment period affects monthly burden; a shorter term with the same loan amount means higher monthly payments and higher default risk.
- **Education and Self_Employed** — proxies for income stability, though weaker signals than the direct financial columns above.

**Less useful features:** Gender and Married status are unlikely to have a genuine causal relationship with repayment ability, and including them risks encoding unfair bias into the model rather than adding real predictive signal.

## Exercise 3 — Training, Evaluating, and Optimizing the Model

**Model choice:** Loan default prediction is a **binary classification problem** (will default / will not default), so an appropriate starting model would be **Logistic Regression** — a simple, interpretable model well-suited to this kind of yes/no outcome. A **Random Forest** or **Decision Tree** classifier would also be reasonable, especially since the data has a mix of categorical (Gender, Education) and numeric (income, loan amount) features, and can also reveal which features matter most — directly answering Exercise 2's question with real evidence, the way our correlation analysis confirmed RAM's dominance in the Mobile Price dataset.

**Steps to evaluate model performance:**
1. **Split the data** into training and test sets (e.g., 80/20), so the model is evaluated on data it has never seen — this prevents the same "makes accuracy look better than it really is" mistake we specifically avoided when we normalized data in the Salary daily challenge.
2. **Check accuracy** — the percentage of correct predictions overall.
3. **Check precision and recall specifically** — this matters a lot here, since the cost of the two mistake types is very different: approving a loan that defaults (a real financial loss) vs. rejecting an applicant who would have repaid (a missed opportunity, but less costly). Precision tells you "of the loans we approved, how many actually repaid"; recall tells you "of all the people who would have repaid, how many did we correctly approve."
4. **F1-score** — a single balanced number combining precision and recall, useful when neither one alone tells the full story.
5. **Confusion matrix** — a simple table breaking down exactly how many correct/incorrect predictions fell into each of the four possible outcomes (true positive, false positive, true negative, false negative) — giving a clearer picture than accuracy alone, especially if defaults are relatively rare in the dataset (an imbalanced classes problem).

## Exercise 4 — Designing Machine Learning Solutions for Specific Problems

**1. Predicting Stock Prices**
→ **Supervised Learning (Regression).** This maps directly onto our Apple stock analysis mini-project — we had years of historical price data with known outcomes (the actual past closing prices), and the goal is to predict a continuous numeric value (a future price) based on patterns in that history, like the moving averages and trends we identified. A real stock-prediction model would be trained the same way: learning from labeled historical data (known past prices) to predict an unknown future value.

**2. Organizing a Library of Books by Genre/Similarity**
→ **Unsupervised Learning (Clustering).** There's no pre-labeled "correct genre" for the model to learn from in advance — the goal is to discover natural groupings among books based on their similarities (word usage, themes, style) without being told the answer upfront. This is conceptually similar to what PCA was doing in this week's Daily Challenge — finding natural structure in the data rather than predicting a known target.

**3. Robot Navigation to Find the Shortest Path in a Maze**
→ **Reinforcement Learning.** The robot doesn't have a labeled dataset of "correct moves" to learn from — instead, it learns through trial and error, receiving a reward (getting closer to the exit) or penalty (hitting a wall, going backward) after each action, gradually improving its strategy over repeated attempts. This is fundamentally different from the other two — there's no fixed dataset at all, just an environment to interact with.

## Exercise 5 — Designing an Evaluation Strategy for Different ML Models

**Supervised Model (Classification) — e.g., the loan default classifier from Exercise 3**
- **Metrics:** Accuracy, Precision, Recall, F1-score (as discussed in Exercise 3) — precision and recall matter especially here since false positives (approving a bad loan) and false negatives (rejecting a good applicant) carry very different real costs.
- **Methods:** Cross-validation — splitting the data into multiple different train/test partitions and averaging the results, rather than relying on just one lucky/unlucky split. ROC curves visualize the tradeoff between correctly identifying defaults vs. incorrectly flagging good applicants, at different decision thresholds.
- **Challenge:** if defaults are rare in the dataset (most loans are repaid), a model could score high "accuracy" just by predicting "no default" every time — this is why precision/recall matter more than raw accuracy alone.

**Unsupervised Model (Clustering) — e.g., the library book-genre grouping from Exercise 4**
- **Metrics:** Silhouette score (measures how well-separated and internally consistent each cluster is), the Elbow method (helps choose the right number of clusters by finding the point where adding more clusters stops meaningfully improving the fit).
- **Challenge:** there's no "correct answer" to check against, unlike supervised learning — evaluation is inherently more subjective, since we're judging whether the discovered groupings make sense, not whether they match a known label.

**Reinforcement Learning Model — e.g., the maze-navigating robot from Exercise 4**
- **Metrics:** Cumulative reward over time (is the robot's total score improving across attempts?), convergence (does its strategy stabilize into a consistently good solution, or does it keep changing erratically?).
- **Key consideration:** the exploration vs. exploitation balance — the robot needs to keep trying new paths (exploration) rather than only repeating the first workable path it found (exploitation), or it risks settling for a mediocre solution instead of discovering the true shortest path.
- **Challenge:** success can take many attempts to become visible, and unlike supervised learning, there's no single "correct" labeled example to check against at any given step — only the accumulated pattern of rewards over time.
