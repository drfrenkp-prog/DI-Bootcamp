```python
import numpy as np

matrix = np.array([[4, 7, 2],
                    [3, 6, 1],
                    [2, 5, 3]])

determinant = np.linalg.det(matrix)
inverse = np.linalg.inv(matrix)

print("Matrix:\n", matrix)
print("Determinant:", determinant)
print("Inverse:\n", inverse)
```

```python
productivity_before = np.random.normal(loc=50, scale=10, size=30)
productivity_after = productivity_before + np.random.normal(loc=5, scale=3, size=30)

mean_before = np.mean(productivity_before)
mean_after = np.mean(productivity_after)
difference = mean_after - mean_before

print(f"Mean productivity before: {mean_before:.2f}")
print(f"Mean productivity after: {mean_after:.2f}")
print(f"Average improvement: {difference:.2f}")

# Hypothesis: H0 = no improvement (mean difference = 0), H1 = training improved productivity (mean difference > 0)
print(f"\nHypothesis: H0 = training had no effect | H1 = training improved productivity")
print(f"Result: since the average improvement ({difference:.2f}) is positive and consistent with the +5 boost built into the data, we'd lean toward rejecting H0")
```

```python
dates_2023 = np.arange('2023-01-01', '2024-01-01', dtype='datetime64[D]')

q1 = dates_2023[(dates_2023 >= np.datetime64('2023-01-01')) & (dates_2023 <= np.datetime64('2023-03-31'))]
q2 = dates_2023[(dates_2023 >= np.datetime64('2023-04-01')) & (dates_2023 <= np.datetime64('2023-06-30'))]
q3 = dates_2023[(dates_2023 >= np.datetime64('2023-07-01')) & (dates_2023 <= np.datetime64('2023-09-30'))]
q4 = dates_2023[(dates_2023 >= np.datetime64('2023-10-01')) & (dates_2023 <= np.datetime64('2023-12-31'))]

print(f"Q1 (Jan-Mar): {len(q1)} days, from {q1[0]} to {q1[-1]}")
print(f"Q2 (Apr-Jun): {len(q2)} days, from {q2[0]} to {q2[-1]}")
print(f"Q3 (Jul-Sep): {len(q3)} days, from {q3[0]} to {q3[-1]}")
print(f"Q4 (Oct-Dec): {len(q4)} days, from {q4[0]} to {q4[-1]}")
```

```python
numpy_array = np.random.randint(1, 50, size=(5, 3))
print("NumPy array:\n", numpy_array)

df_from_array = pd.DataFrame(numpy_array, columns=['X', 'Y', 'Z'])
print("\nConverted to DataFrame:\n", df_from_array)

back_to_numpy = df_from_array.to_numpy()
print("\nConverted back to NumPy array:\n", back_to_numpy)
print("Are they equal?", np.array_equal(numpy_array, back_to_numpy))
```

```python
import matplotlib.pyplot as plt

random_values = np.random.randn(50).cumsum()  # random walk

plt.figure(figsize=(10, 5))
plt.plot(random_values, marker='o', markersize=3, color='steelblue')
plt.title('Line Graph of Random Numbers (Cumulative Random Walk)', fontsize=13, fontweight='bold')
plt.xlabel('Step')
plt.ylabel('Cumulative Value')
plt.grid(alpha=0.3)
plt.show()
```
