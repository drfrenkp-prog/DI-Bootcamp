# Daily Challenge — Real-world Data Analysis Scenario

## Case: Ferrero speeds up decision-making with better data analysis

Source: Setup's 2026 Case Study Showcase (May 14, 2026), presented by Sagepath Reply, Ferrero's analytics partner.

### The situation

Ferrero, the company behind Nutella and Ferrero Rocher, was analyzing its data manually across several separate systems. Every business question meant an analyst had to pull data from different places, clean it, and put it together by hand before anyone could act on it. That took time.

### What changed

Ferrero's analytics partner, Sagepath Reply, brought all of Ferrero's data sources together into one system and built an AI tool called Loop AI to generate insights automatically instead of relying on manual work each time.

### The result

Time to get an answer from the data dropped from days to minutes. Questions that used to take an analyst days to answer could now be answered almost right away.

### What this shows

Without this change, Ferrero would have kept making decisions based on data that was often several days old, simply because pulling it together took that long. This case shows that having data isn't the same as using it well. What mattered here wasn't collecting more data, it was cutting the time between having a question and getting an answer.
