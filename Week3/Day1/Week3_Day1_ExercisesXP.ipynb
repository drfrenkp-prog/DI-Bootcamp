# Exercise 1 — Introduction to Data Analysis

## What is data analysis?

Data analysis is the process of inspecting, cleaning, and interpreting raw
data in order to extract useful information, identify patterns, and support
decision-making. It turns numbers and observations that are hard to act on
into insights a person or a business can actually use.

## Why is data analysis important in modern contexts?

Organizations today generate more data than at any point in history —
website clicks, sales transactions, sensor readings, social media
engagement. Without analysis, that data is just noise. Data analysis is
what converts it into a competitive advantage: it replaces guesswork with
evidence, lets teams catch problems early, and helps predict what's likely
to happen next rather than only explaining what already happened.

## Three areas where data analysis is applied today

**1. Marketing and influencer partnerships.** Campaign performance data —
views, engagement rate, click-through, conversion — is analyzed to decide
which creators and content formats actually drive results, so budget goes
toward what works instead of what looks good on paper.

**2. Healthcare.** Patient data is analyzed to predict disease outbreaks,
personalize treatment plans, and evaluate whether a given treatment is
actually more effective than the alternative.

**3. Finance.** Banks and fintech companies analyze transaction data in
real time to detect fraud, assess credit risk, and inform investment
decisions — often within milliseconds of a transaction occurring.

## Exercise 2 — Dataset Description

This dataset ("Time Americans Spend Sleeping") comes from the U.S. Bureau
of Labor Statistics' American Time Use Survey. Each row represents one
year's average sleep data for a demographic group, with columns for the
year, average hours slept per day, a standard error (margin of
uncertainty), the type of day (e.g. all days vs. weekdays), age group,
the activity being measured (sleeping), and sex. It's a small,
year-by-year summary table rather than individual-level data — good for
spotting trends over time rather than for analyzing individual people.

## Exercise 4 — Qualitative vs. Quantitative Columns

- **Id** — Quantitative (int) — but really just a row number, not a real measurement.
- **SepalLengthCm** — Quantitative (float) — a measured length, can take any decimal value.
- **SepalWidthCm** — Quantitative (float) — same reasoning, a real measurement in cm.
- **PetalLengthCm** — Quantitative (float) — a measured length.
- **PetalWidthCm** — Quantitative (float) — a measured width.
- **Species** — Qualitative (object/string) — a category (e.g. "Iris-setosa"), not a number.

Four of the six columns are true numeric measurements taken with a ruler,
which makes this dataset mostly quantitative — only the flower's species
label is qualitative.

# Exercise 6 — Structured or Unstructured?

- A company's financial reports in Excel — **Structured** (rows/columns, fixed fields)
- Photographs on social media — **Unstructured** (images, no fixed fields)
- A collection of news articles — **Unstructured** (free-form text)
- Inventory data in a relational database — **Structured** (tables with defined columns)
- Recorded interviews from market research — **Unstructured** (raw audio/speech)

# Exercise 7 — Turning Unstructured Data into Structured Data

**Blog posts about travel experiences** — Extract fields like destination,
date, trip length, and rating from each post into a table. The free-text
body can stay as one column, but pulling out those key facts into their
own columns makes the collection sortable and comparable.

**Audio recordings of customer service calls** — Transcribe each call to
text (speech-to-text), then tag each with structured fields: call
duration, issue category, resolved yes/no, and customer sentiment. The
transcript itself is still unstructured text, but the tags around it are
structured.

**Handwritten notes from a brainstorming session** — Manually (or via
OCR) transcribe each idea into a spreadsheet row, with columns for the
idea itself, who proposed it, and a category/theme tag.

**A video tutorial on cooking** — Break it into structured metadata:
recipe name, ingredient list (as a table), step-by-step instructions
with timestamps, and cooking duration — separate from the raw unstructured
video file itself.
