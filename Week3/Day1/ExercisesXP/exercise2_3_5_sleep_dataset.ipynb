## Exercise 2 — Dataset Description

This dataset ("Time Americans Spend Sleeping") comes from the U.S. Bureau
of Labor Statistics' American Time Use Survey. Each row represents one
year's average sleep data for a demographic group, with columns for the
year, average hours slept per day, a standard error (margin of
uncertainty), the type of day (e.g. all days vs. weekdays), age group,
the activity being measured (sleeping), and sex. It's a small,
year-by-year summary table rather than individual-level data — good for
spotting trends over time rather than for analyzing individual people.
