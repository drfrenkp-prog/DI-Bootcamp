## Exercise 4 — Qualitative vs. Quantitative Columns

- **Id** — Quantitative (int) — but really just a row number, not a real measurement.
- **SepalLengthCm** — Quantitative (float) — a measured length, can take any decimal value.
- **SepalWidthCm** — Quantitative (float) — same reasoning, a real measurement in cm.
- **PetalLengthCm** — Quantitative (float) — a measured length.
- **PetalWidthCm** — Quantitative (float) — a measured width.
- **Species** — Qualitative (object/string) — a category (e.g. "Iris-setosa"), not a number.

Four of the six columns are true numeric measurements taken with a ruler,
which makes this dataset mostly quantitative — only the flower's species
label is qualitative.
