# Daily Challenge — Global Weather Trends Analysis

**Scenario:** I'm a data analyst reviewing average monthly temperatures across 10 cities to find seasonal and geographic patterns.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

np.random.seed(42)  # so results are repeatable every time we run this

cities = ['Cairo', 'Oslo', 'Tokyo', 'Sydney', 'Nairobi',
          'Moscow', 'Miami', 'London', 'Delhi', 'Reykjavik']
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun',
          'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']

temps = np.random.uniform(low=-5, high=35, size=(10, 12))

weather_df = pd.DataFrame(temps, index=cities, columns=months)
weather_df.round(1)
```

## Data Analysis 

```python
annual_avg = weather_df.mean(axis=1)
print("Annual average temperature per city:")
print(annual_avg.round(1))

hottest_city = annual_avg.idxmax()
coldest_city = annual_avg.idxmin()

print(f"\nHottest city on average: {hottest_city} ({annual_avg.max():.1f}°C)")
print(f"Coldest city on average: {coldest_city} ({annual_avg.min():.1f}°C)")
```

## Data Visualization

```python
plt.figure(figsize=(14, 7))
for city in weather_df.index:
    plt.plot(weather_df.columns, weather_df.loc[city], marker='o', label=city)

plt.title('Monthly Temperature Trends by City (Synthetic Data)', fontsize=15, fontweight='bold')
plt.xlabel('Month')
plt.ylabel('Temperature (°C)')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left', fontsize=9)
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
import seaborn as sns

plt.figure(figsize=(12, 6))
sns.heatmap(weather_df, annot=True, fmt='.1f', cmap='coolwarm',
            cbar_kws={'label': 'Temperature (°C)'})
plt.title('Monthly Temperature Heatmap by City', fontsize=15, fontweight='bold')
plt.xlabel('Month')
plt.ylabel('City')
plt.tight_layout()
plt.show()
```

## Summary Report

**Hottest city (annual average):** London (17.6°C)
**Coldest city (annual average):** Oslo (9.6°C)

**Observations:**
- The temperature data shows no consistent seasonal pattern across cities — this is expected, since it was generated using `np.random.uniform`, which produces independent random values with no real-world climate logic behind them.
- The heatmap visualization confirms this: rather than smooth warm-summer/cold-winter bands, the colors appear as scattered noise across both cities and months.
- This is a useful reminder that synthetic random data is good for practicing NumPy/Pandas mechanics, but shouldn't be mistaken for realistic climate behavior — real weather data would show clear seasonal cycles and hemisphere-based patterns (e.g. Southern Hemisphere cities being warm in December, not cold).
