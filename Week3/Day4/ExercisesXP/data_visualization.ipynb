## Exercise 1 — Understanding Data Visualization

**Why is data visualization important in data analysis?**
Raw numbers in a table are hard for people to interpret quickly. Visualization turns that data into a picture, so patterns, trends, and outliers that would take a long time to spot manually become obvious at a glance. It also makes it much easier to communicate findings to other people, since a chart is faster to understand than a table of numbers.

**What is the purpose of a line graph?**
A line graph shows how a value changes over a continuous sequence, usually time. It connects data points with a line so you can see the direction and rate of change — whether something is trending up, down, or staying flat.

```python
import matplotlib.pyplot as plt

days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
temperatures = [72, 74, 76, 80, 82, 78, 75]

plt.plot(days, temperatures)
plt.xlabel('Day')
plt.ylabel('Temperature (°F)')
plt.title('Temperature Variation Over the Week')
plt.show()
```

## Exercise 3 — Bar Chart for Monthly Sales

```python
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May']
sales = [5000, 5500, 6200, 7000, 7500]

plt.bar(months, sales)
plt.xlabel('Month')
plt.ylabel('Sales Amount ($)')
plt.title('Monthly Sales')
plt.show()
```

## Exercise 4 — Distribution of CGPA

```python
import matplotlib.pyplot as plt

sns.histplot(data=df, x='What is your CGPA?', color='skyblue')
plt.title('Distribution of Student CGPA')
plt.xlabel('CGPA Range')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.show()
```

## Exercise 5 — Anxiety Levels Across Genders

```python
sns.countplot(data=df, x='Choose your gender', hue='Do you have Anxiety?', palette='Set2')
plt.title('Anxiety Levels by Gender')
plt.xlabel('Gender')
plt.ylabel('Count')
plt.show()
```

## Exercise 6 — Age vs. Panic Attacks

```python
df['PanicAttack_numeric'] = df['Do you have Panic attack?'].map({'Yes': 1, 'No': 0})

sns.scatterplot(data=df, x='Age', y='PanicAttack_numeric')
plt.title('Age vs. Panic Attacks')
plt.xlabel('Age')
plt.ylabel('Panic Attacks (1 = Yes, 0 = No)')
plt.yticks([0, 1])
plt.show()
```

The scatter plot shows two distinct horizontal bands (0 and 1) rather than a smooth spread, since Panic Attack is a binary yes/no answer, not a continuous variable. At most ages, both Yes and No responses appear, suggesting no strong age-based pattern in this sample — ages 21 and 22 show only "No" responses, likely just due to the small sample size at those specific ages.
